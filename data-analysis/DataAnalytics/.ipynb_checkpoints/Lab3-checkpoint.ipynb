```python
import pandas as pd
from scipy import stats
```

```python
# 1. 데이터 불러오기
iris = pd.read_csv("data/iris.csv")

# 2. 품종별로 SepalWidth 데이터 분리
# (컬럼명이 파일에 따라 Sepal.Width 또는 SepalWidth 형태일 수 있으니 확인 필요)
col_name = "SepalWidth" if "SepalWidth" in iris.columns else "Sepal.Width"
groups = [group[col_name].dropna().values for _, group in iris.groupby("Name")]
```

```python
# 3. 일원분산분석(ANOVA) 수행
f_stat, p_val = stats.f_oneway(*groups)

print(f"F-통계량: {f_stat:.4f}")
print(f"p-value: {p_val:.4e}")
```

```python
# 4. 결론 도출 (유의수준 0.05 기준)
if p_val < 0.05:
    print(
        "p-value < 0.05 이므로 귀무가설을 기각합니다. (품종별 SepalWidth에 통계학적으로 유의한 차이가 있음)"
    )
else:
    print(
        "p-value >= 0.05 이므로 귀무가설을 채택합니다. (품종별 SepalWidth에 통계학적으로 유의한 차이가 없음)"
    )
```

```python
# 1. 데이터 불러오기
titanic = pd.read_csv("data/titanic.csv", encoding="euc-kr")
```

```python
# 컬럼명 확인 (대소문자/한글 여부 확인: 예: 'Survived', 'Age' 또는 '생존여부', '나이')
survived_col = "Survived" if "Survived" in titanic.columns else "생존여부"
age_col = "Age" if "Age" in titanic.columns else "나이"
```

```python
# 2. 결측치(NaN) 제거 및 그룹 분리 (t-검정은 결측치가 없어야 함)
survived_age = titanic[titanic[survived_col] == 1][age_col].dropna()
dead_age = titanic[titanic[survived_col] == 0][age_col].dropna()
```

```python
# 3. 등분산성 검정 (Levene test)
levene_stat, levene_p = stats.levene(survived_age, dead_age)
equal_var = levene_p >= 0.05  # 0.05 이상이면 등분산 가정 만족
```

```python
# 4. 독립표본 t-검정 수행
t_stat, p_val = stats.ttest_ind(survived_age, dead_age, equal_var=equal_var)

print(f"t-통계량: {t_stat:.4f}")
print(f"p-value: {p_val:.4f}")
```

```python
# [과제 2번 최종 결론 및 가설 검정]
#
# 1. 가설 설정
# - 귀무가설 (H0): 생존 여부에 따른 승객의 연령 평균에는 차이가 없다. (두 집단의 모평균이 같다)
# - 대립가설 (H1): 생존 여부에 따른 승객의 연령 평균에는 유의미한 차이가 있다. (두 집단의 모평균이 다르다)
#
# 2. 검정 결과
# - t-통계량(t-statistic): -2.0667
# - p-값(p-value): 0.0391
#
# 3. 채택 및 기각 여부 (유의수준 α = 0.05 기준)
# - p-value (0.0391) < 0.05 이므로 유의수준 5% 하에서 귀무가설(H0)을 기각하고, 대립가설(H1)을 채택한다.
#
# 4. 결론
# - 95% 신뢰수준에서 생존자와 사망자 간의 평균 연령에는 통계학적으로 유의미한 차이가 존재한다고 볼 수 있다.
#   (즉, 승객의 생존 여부와 연령은 통계학적 관련성이 있다.)
```
